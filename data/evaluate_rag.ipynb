```python


import os
os.environ["OPENAI_API_KEY"] = "sk-xxxxxxxxxxxxxxxxxxxxxxxxxxxx"  # apna actual key paste karo

import json
from datasets import Dataset
from ragas import evaluate
from ragas.metrics import faithfulness, answer_relevancy, context_precision, context_recall

# Load your JSON file
with open("evaluation_set.json", "r", encoding="utf-8") as f:
    data = json.load(f)

dataset = Dataset.from_list(data)

# Run evaluation
result = evaluate(
    dataset,
    metrics=[faithfulness, answer_relevancy, context_precision, context_recall]
)

# ✅ Overall scores
print("Overall Evaluation Scores:")
print(result)

# ✅ Per-question scores (use result.scores instead of result['results'])
print("\nPer-question Scores:")
for i, row in enumerate(result.scores):
    print(f"Question {i+1}: {data[i]['question']}")
    print("Scores:", row)
```
